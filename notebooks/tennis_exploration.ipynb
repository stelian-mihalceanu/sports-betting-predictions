# Tennis Exploration - ATP & WTA

Target competitions:
- Grand Slams: Australian Open, Roland Garros, Wimbledon, US Open
- ATP Masters 1000
- ATP 500
- WTA 1000
- WTA 500


```python
import pandas as pd
import sys
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT / 'src'))

from data_loader import load_tennis_atp, load_tennis_wta
from data_filters import filter_tennis_target
```

```python
# Load raw data
df_atp = load_tennis_atp()
df_wta = load_tennis_wta()

print(f"ATP matches: {len(df_atp)}")
print(f"WTA matches: {len(df_wta)}")
```

```python
# Filter to target competitions
df_atp_target, df_wta_target = filter_tennis_target(df_atp, df_wta)

print(f"ATP target matches: {len(df_atp_target)}")
print(f"WTA target matches: {len(df_wta_target)}")
```

```python
# Distribution by category
print("\nATP Category Distribution:")
print(df_atp_target['category'].value_counts())

print("\nWTA Category Distribution:")
print(df_wta_target['category'].value_counts())
```

```python
# Surface distribution (if available)
if 'surface' in df_atp_target.columns:
    print("\nATP Surface Distribution:")
    print(df_atp_target['surface'].value_counts())

if 'surface' in df_wta_target.columns:
    print("\nWTA Surface Distribution:")
    print(df_wta_target['surface'].value_counts())
```
